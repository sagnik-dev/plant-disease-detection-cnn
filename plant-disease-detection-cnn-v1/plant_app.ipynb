```python
#this one for EfficientNet


import gradio as gr
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing import image

# 1. Load Model
# Make sure the file path matches exactly where you saved it
model = load_model("best_model_efficientnet.h5")
print("✅ Model loaded!")

# 2. MATCH THE TRAINING SIZE
img_height, img_width = 224, 224  # <--- CHANGED FROM 128 to 224

# 3. Class Names
class_names = [
    'Pepper__bell___Bacterial_spot', 'Pepper__bell___healthy',
    'Potato___Early_blight', 'Potato___Late_blight', 'Potato___healthy',
    'Tomato_Bacterial_spot', 'Tomato_Early_blight', 'Tomato_Late_blight',
    'Tomato_Leaf_Mold', 'Tomato_Septoria_leaf_spot',
    'Tomato_Spider_mites_Two_spotted_spider_mite', 'Tomato__Target_Spot',
    'Tomato__Tomato_YellowLeaf__Curl_Virus', 'Tomato__Tomato_mosaic_virus',
    'Tomato_healthy'
]

def predict_disease(img):
    # A. Resize to 224x224 (Model Requirement)
    img = img.resize((img_width, img_height))

    # B. Convert to Array
    img_array = image.img_to_array(img)

    # C. Add Batch Dimension (1, 224, 224, 3)
    img_batch = np.expand_dims(img_array, axis=0)

    # D. CRITICAL: DO NOT DIVIDE BY 255
    # EfficientNet expects pixels in range [0, 255]
    # img_batch = img_batch / 255.0  <-- I REMOVED THIS LINE

    # E. Predict
    predictions = model.predict(img_batch)
    class_index = np.argmax(predictions)

    result = class_names[class_index]
    confidence = round(100 * np.max(predictions), 2)

    return f"Disease: {result} (Confidence: {confidence}%)"

# Gradio Interface
interface = gr.Interface(
    fn=predict_disease,
    inputs=gr.Image(type="pil"), # Sends a PIL image to the function
    outputs="text",
    title="🌿 Plant Disease Detector (EfficientNet)",
    description="Upload a leaf image. The model uses EfficientNetB0 (224x224) to predict disease."
)

# Launch with share=True for a public link
interface.launch(share=True)
```

```python
#this one is for cnn
import gradio as gr
import numpy as np
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing import image

# Load your trained model
model = load_model("best_model_efficientnet.h5")

# You need to use the same image size as during training
img_height, img_width = 128, 128

# Your class labels (edit this according to your dataset classes)
class_names = [
    'Pepper__bell___Bacterial_spot',
    'Pepper__bell___healthy',
    'Potato___Early_blight',
    'Potato___Late_blight',
    'Potato___healthy',
    'Tomato_Bacterial_spot',
    'Tomato_Early_blight',
    'Tomato_Late_blight',
    'Tomato_Leaf_Mold',
    'Tomato_Septoria_leaf_spot',
    'Tomato_Spider_mites_Two_spotted_spider_mite',
    'Tomato_Target_Spot',
    'Tomato_Tomato_Yellow_Leaf_Curl_Virus',
    'Tomato_Tomato_mosaic_virus',
    'Tomato_healthy'
]
 # and so on...

def predict_disease(img):
    img = img.resize((img_width, img_height))
    img = image.img_to_array(img)
    img = np.expand_dims(img, axis=0)
    img = img / 255.0
    predictions = model.predict(img)
    class_index = np.argmax(predictions)
    result = class_names[class_index]
    confidence = round(100 * np.max(predictions), 2)
    return f"Disease: {result} (Confidence: {confidence}%)"

# Gradio Interface
interface = gr.Interface(
    fn=predict_disease,
    inputs=gr.Image(type="pil"),
    outputs="text",
    title="🌿 Plant Disease Detector",
    description="Upload a leaf image to predict the disease using the trained CNN model."
)

# Launch it!
interface.launch()
```
